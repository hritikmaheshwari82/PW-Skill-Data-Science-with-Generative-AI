

## Question 1: Compare and contrast NLTK and spaCy in terms of features, ease of use, and performance.

**Answer:**

**NLTK (Natural Language Toolkit):**
NLTK is a comprehensive library mainly used for **teaching, research, and prototyping**. It provides a wide range of NLP tools such as tokenization, stemming, lemmatization, parsing, and corpus access.

* **Features:** Very rich, covers almost all classical NLP tasks
* **Ease of Use:** Moderate; requires more manual setup and preprocessing
* **Performance:** Slower, not optimized for production-scale systems

**spaCy:**
spaCy is an **industrial-strength NLP library** designed for production use. It emphasizes speed, efficiency, and modern NLP pipelines.

* **Features:** POS tagging, dependency parsing, NER, word vectors, transformer support
* **Ease of Use:** High; clean API and minimal configuration
* **Performance:** Very fast and optimized for large-scale applications

**Conclusion:**
NLTK is ideal for learning and experimentation, while spaCy is preferred for real-world, performance-critical NLP applications.

## Question 4: Describe the architecture and functioning of a Recurrent Neural Network (RNN).

**Answer:**
A Recurrent Neural Network (RNN) is a type of neural network designed to process **sequential data** such as text or time-series data.

**Architecture:**

* Consists of input, hidden, and output layers
* The hidden layer has a recurrent connection that passes information from the previous time step to the next

**Functioning:**
At each time step, the RNN:

* Takes the current input and the previous hidden state
* Produces a new hidden state
* Generates an output based on the hidden state

This allows RNNs to capture temporal dependencies, making them suitable for NLP tasks like language modeling and sequence prediction. However, standard RNNs suffer from vanishing gradient problems.

## Question 5: What is the key difference between LSTM and GRU networks in NLP applications?

**Answer:**
Both LSTM (Long Short-Term Memory) and GRU (Gated Recurrent Unit) are advanced RNN architectures designed to overcome the limitations of traditional RNNs.

**LSTM:**

* Uses three gates: input, forget, and output gates
* Maintains a separate cell state for long-term memory
* More complex but highly effective for capturing long-range dependencies

**GRU:**

* Uses two gates: update and reset gates
* Combines hidden state and cell state
* Simpler architecture and faster training

**Key Difference:**
LSTM is more expressive and powerful, while GRU is computationally efficient and easier to train. The choice depends on dataset size and performance requirements.

````python
"""Question 7: Given the sample paragraph below, perform string tokenization and
frequency distribution using Python and NLTK:
“Natural Language Processing (NLP) is a fascinating field that combines linguistics,
computer science, and artificial intelligence. It enables machines to understand,
interpret, and generate human language. Applications of NLP include chatbots,
sentiment analysis, and machine translation. As technology advances, the role of NLP
in modern solutions is becoming increasingly critical.”"""

import nltk
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist
import string

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')

# Sample paragraph
text = """
Natural Language Processing (NLP) is a fascinating field that combines linguistics,
computer science, and artificial intelligence. It enables machines to understand,
interpret, and generate human language. Applications of NLP include chatbots,
sentiment analysis, and machine translation. As technology advances, the role of NLP
in modern solutions is becoming increasingly critical.
"""

# Convert text to lowercase
text = text.lower()

# Tokenization
tokens = word_tokenize(text)

# Remove punctuation
tokens = [word for word in tokens if word not in string.punctuation]

# Frequency Distribution
freq_dist = FreqDist(tokens)

# Display frequency distribution
freq_dist
````

````python
"""Question 8: Implement a basic LSTM model in Keras for a text classification task using
the following dummy dataset. Your model should classify sentences as either positive
(1) or negative (0).
# Dataset
texts = [
“I love this project”, #Positive
“This is an amazing experience”, #Positive
“I hate waiting in line”, #Negative
“This is the worst service”, #Negative
“Absolutely fantastic!” #Positive
]
labels = [1, 1, 0, 0, 1]
Preprocess the text, tokenize it, pad sequences, and build an LSTM model to train on
this data. You may use Keras with TensorFlow backend."""

import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Input

# Dataset
texts = [
    "I love this project",                 # Positive
    "This is an amazing experience",       # Positive
    "I hate waiting in line",               # Negative
    "This is the worst service",            # Negative
    "Absolutely fantastic"                 # Positive
]

labels = [1, 1, 0, 0, 1]

# Convert labels to NumPy array (REQUIRED in Keras 3)
labels = np.array(labels)

# Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)
sequences = tokenizer.texts_to_sequences(texts)

# Padding
max_length = 5
padded_sequences = pad_sequences(sequences, maxlen=max_length, padding='post')

# Convert inputs to NumPy array
padded_sequences = np.array(padded_sequences)

# Vocabulary size
vocab_size = len(tokenizer.word_index) + 1

# Build LSTM model (Keras 3 compliant)
model = Sequential([
    Input(shape=(max_length,)),
    Embedding(input_dim=vocab_size, output_dim=16),
    LSTM(32),
    Dense(1, activation='sigmoid')
])

# Compile model
model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

# Model summary
model.summary()

# Train model
model.fit(padded_sequences, labels, epochs=20, verbose=1)

````
